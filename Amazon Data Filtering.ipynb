Took this dataset and filtering it to only have ramen reviews: https://www.kaggle.com/datasets/snap/amazon-fine-food-reviews?resource=download

```python
import pandas as pd
```

```python
!ls
```

```python
df = pd.read_csv("Reviews.csv")
```

```python
df.head()
```

Filtering the csv to only contain reviews about ramen

```python
ramen_df = df[
    df["Text"].str.contains(
        "ramen|instant noodle|cup noodle",
        case=False,
        na=False
    )
]

ramen_df = ramen_df[ramen_df["Text"].str.len() >= 150]

len(ramen_df)
```

```python
ramen_df["Text"].iloc[0]
```

Getting only certain columns, deleting columns such as user id

```python
ramen_df = ramen_df[
    ["ProductId", "Score", "Summary", "Text", "Time",
     "HelpfulnessNumerator", "HelpfulnessDenominator"]
]
```

```python
ramen_df.head(10)
```

```python
ramen_df.to_csv("ramen_reviews_long.csv", index=False)
```
